# Prepare for analysis

## Load packages

```python
from fed.data import (
    clean_gdp_data,
    load_api_gdp_data,
    load_local_gdp_data,
    save_gdp_data,
)
from fed.plot_utils import plot_gdp_trends
```

## Fetch and load the data

```python
## Run this cell only for the first time to get the data from the API.
## After that, you can load the data from local files.

# Load data from API
df_real = load_api_gdp_data(is_real=True)  # Real GDP
df_nominal = load_api_gdp_data(is_real=False)  # Nominal GDP

# Save data to local files
save_gdp_data(df_real, is_real=True)
save_gdp_data(df_nominal, is_real=False)
```

```python
# Load data from local files
df_real = load_local_gdp_data(is_real=True)
df_nominal = load_local_gdp_data(is_real=False)
```

```python
# Show data
df_real.head()
```

```python
# Show data
df_nominal.head()
```

## Clean the data

```python
# Clean data
df_real_clean = clean_gdp_data(df_real)
df_nominal_clean = clean_gdp_data(df_nominal)
```

```python
# Show cleaned data
df_real_clean.head()
```

## Plot the data

```python
# Show plots
plot_gdp_trends(df_real_clean, is_real=True)
```

```python
# Show plots
plot_gdp_trends(df_nominal_clean, is_real=False)
```

### Brief dexcription on the GDP development over time

- **United States (USA)**: Maintains the largest economy by a wide margin, with consistent growth in real and nominal terms throughout the period.
- **China (CHN)**: Experienced rapid growth, becoming the second-largest economy in both real and nominal GDP.
- **Japan (JPN)**: Demonstrates moderate growth in real GDP, with nominal GDP peaking around 2012.
- **United Kingdom (GBR)** and **Germany (DEU)**: Both countries demonstrate moderate growth in real GDP. While their nominal GDPs show more fluctuation, they generally trend upwards.
- **Brazil (BRA)**: While its real GDP shows gradual growth, its nominal GDP peaked around 2011.
- **Switzerland (CHE)**: Represents the smallest economy among the selected countries, but it has shown stable growth in both real and nominal GDP, with very little volatility.
